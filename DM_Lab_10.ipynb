Import Required Libraries

```python
import pandas as pd
import numpy as np
```

Load Dataset Using Pandas

```python
df = pd.read_csv("/content/Retail_Dataset_Cleaned.csv")
print("Dataset Loaded Successfully")
```

```python
df.head()
```

Check Dataset Information

Display Column Names

```python
df.columns
```

Select Revenue Column

```python
df["Revenue"]
```

Calculate Revenue Skewness

```python
revenue_skew = df["Revenue"].skew()
print(revenue_skew)
```

Interpret Revenue Skewness

```python
if revenue_skew > 0:
    print("Positive Skew")
elif revenue_skew < 0:
    print("Negative Skew")
else:
    print("Symmetric Distribution")
```

Calculate Units Sold Skewness

```python
units_skew = df["Units_Sold"].skew()
print(units_skew)
```

Import Visualization Library

```python
import matplotlib.pyplot as plt
```

Plot Revenue Distribution

```python
plt.hist(df["Revenue"], bins=20)
plt.title("Revenue Distribution")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

Plot Units Sold Distribution

```python
plt.hist(df["Units_Sold"], bins=20)
plt.title("Units Sold Distribution")
plt.xlabel("Units Sold")
plt.ylabel("Frequency")
plt.show()
```

Display Final Summary Table

```python
summary = pd.DataFrame({
    "Metric":["Revenue","Units Sold"],
    "Skewness":[revenue_skew,units_skew]
})
summary
```
